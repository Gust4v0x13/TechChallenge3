```python
import pandas as pd
```

```python
df_year_1 = pd.read_csv(
    "../data/raw/survey_year_1/data_2021-2022.csv",
    low_memory=False
)
```

```python
df_year_2 = pd.read_csv(
    "../data/raw/survey_year_2/data_2023-2024.csv",
    low_memory=False
)
```

```python
df_year_3 = pd.read_csv(
    "../data/raw/survey_year_3/data_2025-2026.csv",
    low_memory=False
)
```

```python
datasets = {
    "year_1": df_year_1,
    "year_2": df_year_2,
    "year_3": df_year_3
}
```

```python
for year, dataframe in datasets.items():
    print("=" * 80)
    print(year)

    print("Rows:", dataframe.shape[0])
    print("Columns:", dataframe.shape[1])
    print("Duplicated rows:", dataframe.duplicated().sum())
    print("Missing values:", dataframe.isna().sum().sum())
```
